```python
#write a function to display users and their connections
def display_users(data):
    for user in data["users"]:
        friends = []
        pages = []
        for friend in user["friends"]:
            for user_friend in data["users"]:
                if friend == user_friend["id"]:
                    friends.append(user_friend["name"])

        for page in user["liked_pages"]:
            for user_page in data["pages"]:
                if page == user_page["id"]:
                    pages.append(user_page["name"])
            
        print(f"{user["name"]} is friends with {", ".join(friends)}, liked page {", ".join(pages)}")
```
