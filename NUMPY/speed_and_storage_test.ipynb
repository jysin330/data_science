```python
#using Python list

import numpy as np
import time

size = 1_000_000


l1 = list(range(size))
l2 = list(range(size))


start = time.time()
add = [x+y for x,y in zip(l1, l2)]
print(add[0:10])
end = time.time()
print(f"{end - start} time")
```

```python
#using Numpy Arrays

import numpy as np
import time

size = 1_000_000


l1 = np.array(list(range(size)))
l2 = np.array(list(range(size)))

start = time.time()
add = l1 + l2
end = time.time()
print(add[0:10])
print(f"{end - start} time")
```

```python

l1 = list(range(100))
l2 = np.array(list(range(100)))

# Squaring elements by python list (loop-based)
start_time_list = time.time()
list_squares = [x**2 for x in l1]
end_time_list = time.time()
print(list_squares)
print(start_time_list-end_time_list)

# Squaring elements by numpy array (vectorized)
start_time_numpy = time.time()
numpy_squares = l2**2
end_time_numpy = time.time()
print(numpy_squares)
print(start_time_numpy-end_time_numpy)
```
